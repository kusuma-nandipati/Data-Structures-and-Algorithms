# Approach-2 :FIFO

```python
from collections import deque
queue=deque()
#all possible operation which we can have
dir(queue)
```

```python
queue.appendleft(5)
queue.appendleft(10)
queue.appendleft(15)
queue.appendleft(20)
queue
```

```python
queue.pop()#defaultly removes right side element
```
