```python
#For every case,the overall time complexity is O(n)
class Node:
    def __init__(self,data):
        self.left=None
        self.right=None
        self.data=data
        '''implementation of preorder traversal for a given tree'''
def preOrder(root):
    if root:
        print(str(root.data)+" ",end=' ')
        preOrder(root.left)
        preOrder(root.right)
        '''implementation of inorder traversal for the given tree'''
def inOrder(root):
    if root:
        inOrder(root.left)
        print(str(root.data)+" ",end=' ')
        inOrder(root.right)
        '''implementation of post order traversal for the given tree'''
def postOrder(root):
    if root:
        postOrder(root.left)
        postOrder(root.right)
        print(str(root.data)+" ",end=' ')
#driver code
root=Node(1)
root.left=Node(2)
root.right=Node(3)
root.left.left=Node(4)
root.left.right=Node(5)
print("the preorder traversal is--")
preOrder(root)
print()
print("the inorder traversal is--")
inOrder(root)
print()
print("the postorder traversal is--")
postOrder(root)
```
