```python
from collections import deque
class Node:
    def __init__(self,val):
        self.val=val
        self.left=None
        self.right=None
    def invertTree(self,root):
        if root is not None:
            rightSubTree=self.invertTree(root.right)
            leftSubTree=self.invertTree(root.left)
            root.left=rightSubTree
            root.right=leftSubTree
            return root
        
        #printing inorder of a tree
def levelOrder(root):
    """Print the tree in level order (BFS)."""
    if not root:
        return []
    result = []
    queue = deque([root])
    while queue:
        level_size = len(queue)
        level = []
        for _ in range(level_size):
            node = queue.popleft()
            level.append(node.val)
            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)
        result.append(level)
    return result
        
#driver code
obj=Node(0)
root=Node(4)
root.left=Node(2)
root.right=Node(7)
root.left.left=Node(1)
root.left.right=Node(3)
root.right.left=Node(6)
root.right.right=Node(9)

print("Original tree (in-order):")
res=levelOrder(root)
print(res)

obj = Node(0)  # dummy node to call invertTree
inverted_root = obj.invertTree(root)

print("Inverted tree (in-order):")
reso=levelOrder(inverted_root)
print(reso)
```
