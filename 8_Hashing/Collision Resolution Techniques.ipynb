# 1.Chaining

```python
class HashTable:
    def __init__(self):
        self.Max=10
        self.arr=[[] for i in range(self.Max)]
        
        '''division modulo hash function'''
        
    def hash_function(self,key):
        m=10
        summation=0
        for char in key:
            summation +=ord(char)
        return summation % m
    
    '''modifying setitem method for chaining collision technique'''
    def __setitem__(self,key,value):
        found=False
        h=self.hash_function(key)
        for index,element in enumerate(self.arr[h]):
            if len(element)==2 and element[0]==key:
                self.arr[h][index]=(key,value)
                found=True
                break
        if not found:
            self.arr[h].append((key,value))
            
    '''getting items from hash table'''
    def __getitem__(self,key):
        h=self.hash_function(key)
        for index,element in enumerate(self.arr[h]):
            if element[0]==key:
                return element[1]
            
        '''deleting items from hash table'''
    def __delitem__(self,key):
        h=self.hash_function(key)
        for index,element in enumerate(self.arr[h]):
            if element[0]==key:
                del self.arr[h][index]
                
#driver code
obj=HashTable()
obj['march 6']=123
obj['march 17']=345
#retreive the price at march 6
obj['march 6']
#getting the hashtable
print(obj.arr)
#delete march 17
del obj['march 17']
#hash table after deleting
obj.arr
```

# 1.Linear Probing

```python
class HashTable:
    def __init__(self):
        self.MAX=10
        self.arr=[None for i in range(self.MAX)]
        '''hash function used for linear probing'''
    def hash_function(self,key):
        summation=0
        for char in key:
            summation +=ord(char)
        return summation%self.MAX
    
    '''set item method to add elements to the hash table'''
    def __setitem__(self,key,value):
        h=self.hash_function(key)
        for i in range(self.MAX):
            index=(h+i)%self.MAX
            if self.arr[index] is None:
                self.arr[index]=(key,value)
                return
            elif self.arr[index][0]==key:
                self.arr[index]=(key,value)
                return
        raise Exception("Table is full")
        
        '''get item method to get elements from hash table'''
    def __getitem__(self,key):
        h=self.hash_function(key)
        for i in range(self.MAX):
            index=(h+i)%self.MAX
            if self.arr[index]==None:
                return
            elif self.arr[index][0]==key:
                return self.arr[index][1]
        return None
    
#driver code
obj=HashTable()
#adding keys and values to the hastable
obj["march 2020"]=123
obj["march 2020"]=345
#getting value for a key
obj["march 2020"]
#printing out the hash table
obj.arr
```

# 2.Quadratic Probing

```python
class HashTable:
    def __init__(self):
        self.MAX=10
        self.arr=[None for i in range(self.MAX)]
        '''set item method to add elements to the hash table'''
    def __setitem__(self,key,value):
        h=key%self.MAX
        c1,c2=1,1
        for i in range(self.MAX):
            index=(h+c1*i+c2*i**2)%self.MAX
            if self.arr[index] is None:
                self.arr[index]=(key,value)
                return
            elif self.arr[index][0]==key:
                self.arr[index]=(key,value)
                return
        print("table is full")
        '''method to get elements from table'''
    def __getitem__(self,key):
        h=key%self.MAX
        for i in range(self.MAX):
            index=(h+i)%self.MAX
            if self.arr[index]==None:
                return
            elif self.arr[index][0]==key:
                return self.arr[index][1]
        return None
    
#driver code
obj=HashTable()
#adding keys and values to the hastable
obj[50]=50
obj[75]=75
obj[99]=99
obj[20]=20
obj[35]=35
obj[88]=88
obj[45]=45
obj[23]=23
obj[55]=55
obj[67]=67
#getting value for a key
obj[23]
#printing out the hash table
print(obj.arr)
```

# Double Hashing 

```python
class HashTable:
    def __init__(self):
        self.MAX=10
        self.arr=[None for i in range(self.MAX)]
        '''set item method to add elements to the hash table'''
    def __setitem__(self,key,value):
        h1=key%self.MAX
        h2=1+key %(self.MAX-2)
        for i in range(self.MAX):
            index=(h1+i*h2)%self.MAX
            if self.arr[index] is None:
                self.arr[index]=(key,value)
                return
            elif self.arr[index][0]==key:
                self.arr[index]=(key,value)
                return
        print(f"Table is full! Could not insert {key}")

        '''method to get elements from table'''
    def __getitem__(self,key):
        h1=key%self.MAX
        h2= 1+key %(self.MAX-2)
        for i in range(self.MAX):
            index=(h1+i*h2)%self.MAX
            if self.arr[index]==None:
                return
            elif self.arr[index][0]==key:
                return self.arr[index][1]
        return None
    
    
#driver code
obj=HashTable()
#adding keys and values to the hastable
obj[50]=50
obj[75]=75
obj[99]=99
obj[20]=20
obj[35]=35
obj[88]=88
obj[45]=45
obj[23]=23
obj[55]=55
obj[67]=67
#getting value for a key
obj[23]
#printing out the hash table
print(obj.arr)
```
