```python
import heapq

class Node:
    def __init__(self, char, freq):
        self.char = char
        self.freq = freq
        self.left = None
        self.right = None

    def __lt__(self, other):
        return self.freq < other.freq


def build_huffman(freqs):
    heap = [Node(ch, fr) for ch, fr in freqs.items()]
    heapq.heapify(heap)

    while len(heap) > 1:
        left = heapq.heappop(heap)
        right = heapq.heappop(heap)

        parent = Node(None, left.freq + right.freq)
        parent.left = left
        parent.right = right

        heapq.heappush(heap, parent)

    return heap[0]


def generate_codes(root):
    codes = {}

    def traverse(node, path):
        if node.char is not None:
            codes[node.char] = path
            return
        traverse(node.left, path + "0")
        traverse(node.right, path + "1")

    traverse(root, "")
    return codes


# frequencies given in the question
freqs = {"a":45, "b":15, "c":2, "d":30, "e":5, "f":3}

root = build_huffman(freqs)
codes = generate_codes(root)

print("Huffman Codes:")
for ch in sorted(codes):
    print(ch, ":", codes[ch])
```
