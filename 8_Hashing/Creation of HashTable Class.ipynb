# Creation of hash table using division modulo hash function

```python
class HashTable:
    def __init__(self):
        self.Max=10
        self.arr=[None for i in range(self.Max)]
    def hash_function(self,key):
        summation=0
        m=10
        for char in key:
            summation +=ord(char)
        return summation%m
    def addItem(self,key,value):
        h=self.hash_function(key)
        self.arr[h]=value
    def getItem(self,key):
        h=self.hash_function(key)
        return self.arr[h]
    
#driver code
obj=HashTable()
obj.addItem('March 2024',234)
print(obj.getItem('March 2024'))
print(obj.arr)
```

# Creation of hash table using mid square hash function

```python
class HashTable:
    def __init__(self):
        self.Max=10
        self.arr=[None for i in range(self.Max)]
        
    #hash function method definition
    def hash_function(self,key):
        m=10
        summation=0
        for char in key:
            summation +=ord(char)
        squared=summation**2
        squared_str=str(squared)
        mid=len(squared_str)//2
        if len(squared_str)%2==0:
            mid_digits=squared_str[mid-1:mid+1]
        else:
            mid_digits=squared_str[mid]
        hash_value=int(mid_digits)%m
        return hash_value
    
    #adding item into hash table
    def addItem(self,key,value):
        h=self.hash_function(key)
        self.arr[h]=value
        
    #getting item from hash table
    def getItem(self,key):
        h=self.hash_function(key)
        return self.arr[h]
    
#driver code
obj=HashTable()
obj.addItem('March 2024',234)#goes to 7th index
obj.addItem('March 2025',999)#goes to 1st index
obj.addItem('MMMMMMMdT',100)#collision occured,rewrites the 234 at 7th index
print(obj.getItem('March 2025'))
print(obj.arr)
```
