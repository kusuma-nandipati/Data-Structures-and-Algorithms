```python
from collections import defaultdict,deque
class Solution:
    def validPath(self,n,edges,source,destination):
        graph=defaultdict(list)
        for a,b in edges:
            graph[a].append(b)
            graph[b].append(a)
        #BFT
        visited=[False]*n
        visited[source]=True
        queue=deque([source])
        while queue:
            node=queue.popleft()
            #To check whether direct path exists or not
            if node==destination:
                return True
            
            #check for adjacent_nodes
            for adjacent_node in graph[node]:
                if not visited[adjacent_node]:
                    visited[adjacent_node]=True
                    queue.append(adjacent_node)
        #queue got empty--->return False
        return False
    
    
#driver code
obj=Solution()
if obj.validPath(3,[[0,1],[1,2],[2,0]],0,2):
    print("yes ,there exists a valid path")
else:
    print("there doesnt exists a valid path")
```
