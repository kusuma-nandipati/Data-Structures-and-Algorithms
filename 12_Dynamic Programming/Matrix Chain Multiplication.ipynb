# 1. Recursive Approach

```python
import sys
def MatrixChainOrder(p,i,j):
    #base case condition
    if i==j:
        return 0
    min_val=sys.maxsize
    #k value--range from 1 to 3 for 4 matrices
    for k in range(i,j):
        count=(MatrixChainOrder(p,i,k)+MatrixChainOrder(p,k+1,j)+p[i-1]*p[k]*p[j])
        #updating min value
        if count<min_val:
            min_val=count
    return min_val

#driver code
if __name__=="__main__":
    arr=[1,3,1,2,3]
    n=len(arr)
    print(MatrixChainOrder(arr,1,n-1))
```

# 3.Tabulation

```python
import sys
def MatrixMultiplication(p,n):
    dp=[[0 for _ in range(n)]for _ in range(n)]
    #L-- length of the chain--2,3,4 for 4 matrices
    for L in range(2,n):
        for i in range(1,n-L+1):
            j=i+L-1
            dp[i][j]=sys.maxsize
            for k in range(i,j):
                cost=dp[i][k]+dp[k+1][j]+p[i-1]*p[k]*p[j]
                dp[i][j]=min(dp[i][j],cost)
    return dp[1][n-1]
    
    
    
#driver code
arr=[1,3,1,2,3]
n=len(arr)
print(MatrixMultiplication(arr,n))
```
