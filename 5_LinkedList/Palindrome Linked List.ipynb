# Approach-1 :Using Stack 
#T.c:O(n), S.C:O(n)

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
        
class LinkedList:
    def __init__(self):
        self.head=None
    def insertionAtBeginning(self,new_data):
        new_node=Node(new_data)
        temp=self.head
        new_node.next=self.head
        self.head=new_node
        
    def printList(self):
        temp=self.head
        while temp:
            print(str(temp.data)+" ",end=" ")
            temp=temp.next
    #method for palindrome
    def isPalindrome(self):
        temp=self.head
        current_node=temp
        s=[]
        while current_node is not None:
            s.append(current_node.data)
            current_node=current_node.next
        while temp is not None:
            c=s.pop()
            if temp.data !=c:
                return False
            temp=temp.next
        return True
            
#driver code
llist=LinkedList()
llist.insertionAtBeginning(10)
llist.insertionAtBeginning(20)
llist.insertionAtBeginning(30)
llist.insertionAtBeginning(20)
llist.insertionAtBeginning(10)
llist.printList()
if llist.isPalindrome():
    print("palindrome")
else:
    print("not a palindrome")
```

# Using Recursion
#T.C:O(n),s.c:O(n)

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
        
class LinkedList:
    def __init__(self):
        self.head=None
    def insertionAtBeginning(self,new_data):
        new_node=Node(new_data)
        temp=self.head
        new_node.next=self.head
        self.head=new_node
        
    def printList(self):
        temp=self.head
        while temp:
            print(str(temp.data)+" ",end=" ")
            temp=temp.next
    #method for palindrome
    def isPalindrome(self):
        temp=self.head
        start=[temp]
        return self.isPalindromeRecur(temp,start)
    def isPalindromeRecur(self,end,start):
        if end is None:
            return True
        right=self.isPalindromeRecur(end.next,start)
        ans=right and start[0].data==end.data
        start[0]=start[0].next
        return ans
        
            
#driver code
llist=LinkedList()
llist.insertionAtBeginning(10)
llist.insertionAtBeginning(20)
llist.insertionAtBeginning(30)
llist.insertionAtBeginning(80)
llist.insertionAtBeginning(10)
llist.printList()
if llist.isPalindrome():
    print("palindrome")
else:
    print("not a palindrome")
```

# using Iterative Approach(using fast and slow pointers)
T.C:O(n),S.C:O(1)

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
        
class LinkedList:
    def __init__(self):
        self.head=None
    def insertionAtBeginning(self,new_data):
        new_node=Node(new_data)
        temp=self.head
        new_node.next=self.head
        self.head=new_node
        
    def printList(self):
        temp=self.head
        while temp:
            print(str(temp.data)+" ",end=" ")
            temp=temp.next
    #method for palindrome
    def isPalindrome(self):
        temp=self.head
        if temp is None or temp.next is None:
            return True
        slow,fast=temp,temp
        while fast.next and fast.next.next:
            slow=slow.next
            fast=fast.next.next
        head2=self.reversal(slow.next)
        slow.next=None
        ret=self.isIdentical(temp,head2)
        head2=self.reversal(head2)
        slow.next=head2
        return ret
    def reversal(self,head):
        prev=None
        nextnode=None
        current=head
        while current:
            nextnode=current.next
            current.next=prev
            prev=current
            current=nextnode
        return prev
    def isIdentical(self,n1,n2):
        while n1 and n2:
            if n1.data !=n2.data:
                return False
            n1=n1.next
            n2=n2.next
        return True
        
            
#driver code
llist=LinkedList()
llist.insertionAtBeginning(10)
llist.insertionAtBeginning(20)
llist.insertionAtBeginning(30)
llist.insertionAtBeginning(20)
llist.insertionAtBeginning(10)
llist.printList()
if llist.isPalindrome():
    print("palindrome")
else:
    print("not a palindrome")
```
