```python
#Fibonacci Series using recursion
def fib(n):
    if n==0 or n==1:
        return n
    else:
        result = fib(n-1)+fib(n-2)
    return result
#driver code 
#function calling
fib(7)
```

```python
fib(50)
```

Observation :
    when the n value is small,the approach is taking  a few seconds to generate output,
    But,when the n value is quite high,the above recursive approach is taking too long time because of exponential time complexity.
    
 Why it is taking too much time when n is quite high?

Overlapping subproblem - Recomputation of same subproblems again and again.

Solution of above problem(Overlapping subproblem):

Dynamic Problem
1.Memoization(Top Down Approach)
2.Tabulation(Bottom Up Approach)
    

# 1.Memoization


With using recursion only..storing the result of every recursive call in a data structure.

T.C:O(n) 


S.C:O(n)

```python
def fib_topDown(n,memo):
    if memo[n] is not None:
        return memo[n]
    if n==0 or n==1:
        result=n
    else:
        result=fib_topDown(n-1,memo)+fib_topDown(n-2,memo)
    memo[n]=result
    return result

def fib_memo(n):
    memo=[None]*(n+1)
    return fib_topDown(n,memo)
```

```python
fib_memo(7)
```

```python
fib_memo(50)
```

```python
fib_memo(100)
```

```python
fib_memo(1000)
```

```python
fib_memo(10000)
```

Tabulation: 
idea: The problem is mainly because of recursion,so..Get rid of recursion itself

Time Complexity: O(n)

Space Complexity: O(n)

```python
def fib_bottomUp(n):
    bottomUp=[None]*(n+1)
    bottomUp[0]=0
    bottomUp[1]=1
    for i in range(2,n+1):
        bottomUp[i]=bottomUp[i-1]+bottomUp[i-2]
    return bottomUp[n]

```

```python
fib_bottomUp(7)
```

```python
fib_bottomUp(50)
```

```python
fib_bottomUp(10000)
```

```python
fib_bottomUp(5000)
```

Note: complexity wise both memoization(top to bottom) and tabulation(bottom to top) are same,but there exists logical difference..

# Highly Optimized Approach for fibonacci series

```python
def fib(n):
    first=0
    second=1
    for i in range(2,n+1):
        nextv=first+second
        first,second=second,nextv
    return nextv
fib(7)
```

```python
fib(1000)
```

```python
fib(5000)
```
