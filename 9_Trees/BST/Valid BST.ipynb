```python
class Node:
    def __init__(self,data):
        self.data=data
        self.left=None
        self.right=None
    def inOrder(self,root,inarr):
        if root:
            self.inOrder(root.left,inarr)
            inarr.append(root.data)
            self.inOrder(root.right,inarr)
        return inarr
    def validBST(self,root):
        inarr=[]
        self.inOrder(root,inarr)
        for i in range(1,len(inarr)):
            if inarr[i]<=inarr[i-1]:
                return False
        return True
    
    
obj=Node(0)    
root=Node(5)
root.left=Node(1)
root.right=Node(3)
if obj.validBST(root):
    print("this is valid BST")
else:
    print("not a valid BST")
```
