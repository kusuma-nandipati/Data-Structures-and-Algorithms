```python
import heapq
def job_sequencing_heap(arr):
    #sort in ascending using deadline
    arr.sort(key=lambda x:x[2])
    min_heap=[]
    for job in arr:
        job_id,profit,deadline=job
        if len(min_heap)<deadline:
            heapq.heappush(min_heap,(profit,job_id))
        elif min_heap[0][0]<profit:
            heapq.heappop(min_heap)
            heapq.heappush(min_heap,(profit,job_id))
    result=sorted(min_heap,key=lambda x:-x[0])
    print("selected jobs: \n")
    for p,q in result:
        print(q,p)
    print("max profit: \n",sum(p for p,q in result))
    
    
    
    
#driver code
arr=[['j1',55,5],
     ['j2',65,2],
     ['j3',75,7],
     ['j4',60,3],
     ['j5',70,2],
     ['j6',50,1],
     ['j7',85,4],
     ['j8',68,5],
     ['j9',45,3]]
job_sequencing_heap(arr)
```
