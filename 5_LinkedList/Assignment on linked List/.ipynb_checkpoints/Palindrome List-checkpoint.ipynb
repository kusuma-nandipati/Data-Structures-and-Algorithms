# Approach-1 :Using Stack

```python
#T.C:O(n)
#S.C:O(n)
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
class LinkedList:
    def __init__(self):
        self.head=None
        '''inserting the elements into the list'''
    def insertAtpos(self,index,new_data):
        new_node=Node(new_data)
        count=0
        temp=self.head
        if index==0:
            new_node.next=self.head
            self.head=new_node
            return
        while temp is not None and count<index-1:
            temp=temp.next
            count +=1
        if temp is None:
            print("index out of range")
            return
        new_node.next=temp.next
        temp.next=new_node
        
        
        '''printing out the list'''
    def printList(self):
        temp=self.head
        while temp:
            print(str(temp.data)+" ",end=" ")
            temp=temp.next
        
    def isPalindrome(self):
        temp=self.head
        current_node=temp
        s=[]
        while current_node is not None:
            s.append(current_node.data)
            current_node = current_node.next
        while temp is not None:
            c = s.pop()
            if temp.data != c:
                return False
            temp = temp.next

        return True
    
#driver code
llist=LinkedList()
llist.insertAtpos(0,10)
llist.insertAtpos(1,20)
llist.insertAtpos(2,30)
llist.insertAtpos(3,30)
llist.insertAtpos(4,20)
llist.insertAtpos(5,10)

llist.printList()
if llist.isPalindrome():
    print("The list is palindrome")
else:
    print("The list is not palindrome")
        
```
