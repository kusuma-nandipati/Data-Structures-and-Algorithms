# searching a node by data value

# T.C:O(n)

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
class LinkedList:
    def __init__(self):
        self.head=None
        
        
    '''inserting elements into the list'''
    
    
    def insertAtpos(self,index,new_data):
        new_node=Node(new_data)
        temp=self.head
        count=0
        #insertion at beginning
        if index==0:
            new_node.next=self.head
            self.head=new_node
            return
        while temp is not None and count<index-1:
            temp=temp.next
            count +=1
        if temp==None:
            print("index out of range")
            return
        new_node.next=temp.next
        temp.next=new_node
            
  ''' printing out the Linked List'''

    def printList(self):
        temp=self.head
        while temp:
            print(str(temp.data)+" ",end=" ")
            temp=temp.next
            
            
    '''function to search an element in the Linked List'''
    
    
    def searchNode(self,key):
        temp=self.head
        while temp is not None:
            if temp.data==key:
                return True
            temp=temp.next
        return False
            
            
            
#driver code
llist=LinkedList()
llist.insertAtpos(0,10)
llist.insertAtpos(1,20)
llist.insertAtpos(2,30)
llist.insertAtpos(3,40)
llist.printList()
res=llist.searchNode(10)
if res:
    print("search key found")
else:
    
    print("search key not found")
```
