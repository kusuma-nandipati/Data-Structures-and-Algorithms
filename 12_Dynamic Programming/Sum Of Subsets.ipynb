# 1.Recursive Approach

```python
def isSubsetSum(Set,n,Sum):
    if Sum==0:
        return True
    if n==0:
        return False
    if Set[n-1]>Sum:
        return isSubsetSum(Set,n-1,Sum)
    return (isSubsetSum(Set,n-1,Sum))or (isSubsetSum(Set,n-1,Sum-Set[n-1]))
    
    
    
#driver code
Set=[3,34,4,12,5,5]
Sum=38
n=len(Set)
if isSubsetSum(Set,n,Sum):
    print("Found a Subset with given sum")
else:
    print("No subset with given Sum")
```

# 2.Memoization

```python
def isSubsetSum(arr,n,target,dp):
    #base cases
    if target==0:
        return True
    if n==0:
        return False
    #check dp table if already solved
    if dp[n][target] !=-1:
        return dp[n][target]
    #if the current array element is > target value,skip that element
    if arr[n-1]>target:
        dp[n][target]=isSubsetSum(arr,n-1,target,dp)
        return dp[n][target]
    #for computing --follow include exclude
    include=isSubsetSum(arr,n-1,target-arr[n-1],dp)
    exclude=isSubsetSum(arr,n-1,target,dp)
    dp[n][target]=include or exclude
    #return if the subset for the target is found
    return dp[n][target]

def SubsetSum(arr,target):
    n=len(arr)
    dp=[[-1]*(target+1) for i in range(n+1)]
    print(isSubsetSum(arr,n,target,dp))
    
#driver code
arr=[3,34,4,12,5,2]
target=1
SubsetSum(arr,target)
```

# 3.Tabulation

```python
def isSubsetSum(Set,n,Sum):
    Subset=([[False for _ in range(Sum+1)]for _ in range(n+1)])
    #if sum is 0
    for i in range(n+1):
        Subset[i][0]=True
    
    #if no element is selected,but sum is there-->False
    for j in range(1,Sum+1):
        Subset[0][j]=False
    
    #compute table values by include and exclude by using recursion
    for i in range(1,n+1):
        for j in range(1,Sum+1):
            if j<Set[i-1]:
                Subset[i][j]=Subset[i-1][j]
            if j>=Set[i-1]:
                Subset[i][j]=(Subset[i-1][j] or Subset[i-1][j-Set[i-1]])
    #printing a clean dp table
    #<4,<3 demoted width between cells for a clean table
    print("Table is--\n")
    print("      ", end="")
    for j in range(Sum+1):
        print(f" j={j:<4}", end="")
    print()

    for i in range(n+1):
        print(f"i={i:<3}  ", end="")
        for j in range(Sum+1):
            print(f"{Subset[i][j]:<7}", end="")
        print()
    #return the result
    return Subset[n][Sum]
    
    
    
#driver code
if __name__=="__main__":
    Set=[3,4,5,2]
    Sum=9
    n=len(Set)
    if isSubsetSum(Set,n,Sum)==True:
        print("Found a Subset that sums to target")
    else:
        print("not found")
```
