# Implementation of Stack using List(Dynamic List)

```python
#Insertion (push operation)
#T.C for insertion:O(1)
stack=[]
stack.append('https://google.com/')
stack.append('https://leetcode.com/')
stack.append('https://leetcode.com/problem set/all/')
```

```python
#POP Operation
#T.C:O(1)
stack.pop()
```

```python
'''printing out the stack after pop operation'''
stack
```

# Implementation of Stack using deque

```python
from collections import deque
stack=deque()
```

```python
stack.append('https://google.com/')
stack.append('https://leetcode.com/')
stack.append('https://leetcode.com/problem set/all/')
stack
```

```python
stack.pop()
```
