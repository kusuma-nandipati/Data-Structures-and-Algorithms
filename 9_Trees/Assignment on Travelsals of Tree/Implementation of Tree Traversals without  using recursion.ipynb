# Pre-Order Traversal

```python
#preorder---root,left,right
class Node:
    def __init__(self,data):
        self.left=None
        self.right=None
        self.data=data
def preOrder(root):
    if root is None:
        return
    stack=[]
    #append the root node in the stack
    stack.append(root)
    #first push right root node and then left,so that we get left element as popped one
    while len(stack)>0:
        popped=stack.pop()
        print(str(popped.data)+" ",end=" ")
        if popped.right:
            stack.append(popped.right)
        if popped.left:
            stack.append(popped.left)
        
        
    
#driver code
root=Node(10)
root.left=Node(20)
root.right=Node(30)
root.left.left=Node(40)
root.left.right=Node(50)
root.right.left=Node(60)
root.right.right=Node(70)
preOrder(root)

        
```

# PostOrder Traversal

```python
def postOrder(root):
    current=root
    stack=[]
    new_stack=[]
    stack.append(current)
    while stack:
        current=stack.pop()
        new_stack.append(current.data)
        if current.left:
            stack.append(current.left)
        if current.right:
            stack.append(current.right)
    while new_stack:
        i=new_stack.pop()
        print(i,end=" ")
        
print("postorder traversal--")
postOrder(root)
print()
```

# Inorder Traversal

```python
def inOrder(root):
    current=root
    stack=[]
    while(1):
        if current is not None:
            stack.append(current)
            current=current.left
        elif len(stack)>0:
            current=stack.pop()
            print(str(current.data)+" ",end=' ')
            current=current.right
        else:
            break
    
print("The Inorder Traversal is---")
inOrder(root)
```
