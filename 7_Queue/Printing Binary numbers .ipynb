```python
from collections import deque
class Queue:
    def __init__(self):
        self.container=deque()
    def enqueue(self,val):
        return self.container.appendleft(val)
    def dequeue(self):
        return self.container.pop()
    def front(self):
        return self.container[-1]
#method definition
def produceBinaryNum(n):
    q=Queue()
    q.enqueue("1")
    for i in range(n):
        front=q.front()
        print(front,end=" ")
        q.enqueue(front+"0")
        q.enqueue(front+"1")
        q.dequeue()
#driver code
n=10
result=produceBinaryNum(n)
```
