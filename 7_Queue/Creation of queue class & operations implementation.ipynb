```python
from collections import deque
class Queue:
    def __init__(self):
        self.container=deque()
    def enqueue(self,x):
        return self.container.appendleft(x)
    def dequeue(self):
        return self.container.pop()
    def is_empty(self):
        return len(self.container)==0
    def size(self):
        return len(self.container)
```

```python
#driver code
#creating object of Queue class
q=Queue()
```

```python
q.enqueue({
    'company':'Reliance',
    'timestamp':'05-11-2022 16:38',
    'price':142.78
})
q.enqueue({
    'company':'Reliance',
    'timestamp':'05-11-2022 16:38',
    'price':147.78
})
q.enqueue({
    'company':'Reliance',
    'timestamp':'05-11-2022 16:38',
    'price':150.78
})
```
