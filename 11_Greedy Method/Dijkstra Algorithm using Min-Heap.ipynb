```python
import heapq
def calculate_distances(graph,starting_vertex):
    #intiliazing all nodes with infinity except starting node
    distances={vertex: float('infinity') for vertex in graph}
    distances[starting_vertex]=0
    #min heap
    pq=[(0,starting_vertex)]
    #pushing into minheap-->when at present distance>previous distance
    while len(pq)>0:
        current_distance,current_vertex=heapq.heappop(pq)
        if current_distance > distances[current_vertex]:
            continue
            #calculating at present distance-->popped node distance+ weight
        for neighbour,weight in graph[current_vertex].items():
            distance=current_distance+ weight
            if distance<distances[neighbour]:
                distances[neighbour]=distance
                heapq.heappush(pq,(distance,neighbour))
    return distances
#given graph   
    
    
distances={
    'A':{'B':2,'C':5,'D':2,'E':7,'F':50},
    'B':{'C':2,'D':1,'E':2,'F':60},
    'C':{'B':3,'E':2,'F':90},
    'D':{'E':1,'F':3},
    'E':{'D':4,'F':4},
    'F':{}
}
print(calculate_distances(distances,'A'))
```
