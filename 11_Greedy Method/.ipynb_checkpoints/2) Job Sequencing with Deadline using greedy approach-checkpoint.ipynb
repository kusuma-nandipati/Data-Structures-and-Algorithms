```python
#Every job needs to be completed before or equal to the given deadline
## Optimization: Maximum Profit
#using greedy approach
def job_sequencing(arr,max_Deadline):
    n=len(arr)
    #sort the given data ,with repect to maximum profit in descending order
    arr.sort(key=lambda x:x[1] ,reverse=True)
    #initialize result and job arrays to the length of max_Deadline
    result=[False]*max_Deadline
    job=[-1]*max_Deadline
    
    #for each entry from the given data,schedule the jobs
    for i in range(n):
        for j in range(min(max_Deadline-1,arr[i][2]-1),-1,-1):
            if result[j]is False:
                result[j]=True
                job[j]=arr[i][0]
                break
                
                
    max_profit = 0
    for j_id in job:
        if j_id != -1:  # if a job is scheduled
            # find this job in arr and add its profit
            for item in arr:
                if item[0] == j_id:
                    max_profit += item[1]
                    break
    print("the job sequence is--",job)
    print("max profit--",max_profit)
    
#driver code
arr=[['j1',55,5],
     ['j2',65,2],
     ['j3',75,7],
     ['j4',60,3],
     ['j5',70,2],
     ['j6',50,1],
     ['j7',85,4],
     ['j8',68,5],
     ['j9',45,3]]
max_Deadline=max(deadline[2] for deadline in arr)
job_sequencing(arr,max_Deadline)
```
