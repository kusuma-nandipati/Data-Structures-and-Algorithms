# 1.Memoization

```python
def knapsack(wt,val,W,n,t):
    #base case condition
    if n==0 or W==0:
        return 0
    #checking is previously performed the recursive call
    if t[n][W] !=-1:
        return t[n][W]
    '''recursive conditions'''
    if wt[n-1] <=W:
        t[n][W]=max(val[n-1]+knapsack(wt,val,W-wt[n-1],n-1,t),knapsack(wt,val,W,n-1,t))
        return t[n][W]
    # entirely skip the element is exceeds maximum capacity
    elif wt[n-1]>W:
        t[n][W]=knapsack(wt,val,W,n-1,t)
        return t[n][W]
    
#driver code
val=[60,100,120]
wt=[10,20,30]
W=50
n=len(val)
#initializing memoization table for storing recursive calls
t=[[-1 for i in range(W+1)] for j in range(n+1)]
print(knapsack(wt,val,W,n,t))
```

# 2.Tabulation

```python
def knapsack(W,wt,val,n):
    #table intialization
    k=[[0 for i in range(W+1)] for j in range(n+1)]
    for i in range(n+1):
        for W in range(W+1):
            #base case condition
            if i==0 or W==0:
                k[i][W]=0
            elif wt[i-1] <=W:
                #including the object and doesnt include the object
                k[i][W]=max(val[i-1]+k[i-1][W-wt[i-1]],k[i-1][W])
                
            else:
                #wt[i-1]>W --->skip the current element and copy previous element into it 
                k[i][W]=k[i-1][W]
    return k

def contributing_objects_by_profit(val,W,n,table):
    arr=[0]*n
    net_profit=table[n][W]
    w=W
    '''remember'''#here iterations denotes the object number
    for i in range(n,0,-1):
        if table[i][w] !=table[i-1][w]:
            arr[i-1]=1
            net_profit -=val[i-1]#updating the net profit
            for new_w in range(w - wt[i-1], -1, -1):#finding out the new weight in the above row 
                if table[i-1][new_w] == net_profit:#checking if the current profit is came from above row or not
                    w = new_w
                    break
    return arr
                
        
def contributing_objects_by_weight(val,W,n,table):
    arr=[0]*n
    res=table[n][W]
    w=W
    for i in range(n,0,-1):
        if res !=table[i-1][w]:# checking current profit with the above profit
            arr[i-1]=1
            res -=val[i-1]
            w -=wt[i-1]# updating the weight
    return arr

#Driver code
val=[1,2,3]
wt=[4,5,1]
W=4
n=len(val)
table=knapsack(W,wt,val,n)
print("finding out The objects contributed to the max profit is by profit updating method--\n")
print(contributing_objects_by_profit(val,W,n,table))
print("finding out The objects contributed to the max profit is by weight updating method--\n")
print(contributing_objects_by_weight(val,W,n,table))
print("table---\n")
print(table)
            
```
