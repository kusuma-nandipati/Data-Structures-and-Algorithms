```python
class MyQueue:
    def __init__(self):
        self.stack1=[]
        self.stack2=[]
    def push(self,x):
        self.stack1.append(x)
    def pop(self):
        if len(self.stack1)==0 and len(self.stack2)==0:
            return -1
        while len(self.stack1)>1:
            num=self.stack1.pop()
            self.stack2.append(num)
        our_num=self.stack1.pop()
        while len(self.stack2)>0:
            num=self.stack2.pop()
            self.stack1.append(num)
        return our_num
    def peek(self):
        if len(self.stack1)==0 and len(self.stack2)==0:
            return -1
        
        return self.stack1[0]
            
    
    
    
    
#driver code
obj=MyQueue()
obj.push(1)
obj.push(2)
obj.push(3)
obj.push(4)
pop_element=obj.pop()
print("popped element-----",pop_element)
peek_element=obj.peek()
print("peek element------",peek_element)
```
