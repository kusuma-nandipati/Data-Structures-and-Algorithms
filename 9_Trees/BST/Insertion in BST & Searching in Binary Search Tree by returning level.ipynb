```python
class Node:
    #create a class of node of node in a binary search tree
    def __init__(self,data):
        self.data=data
        self.left=None
        self.right=None
        #insertion of a node in a binary search tree
def insertBST(root,key):
    #if the node is empty,insert at that place 
    if root is None:
        return Node(key)
    else:
        if root.data==key:
            return root
        elif root.data<key:
            #recursion call for right subtree
            root.right=insertBST(root.right,key)
        else:
            #recursion call for left subtree
            root.left=insertBST(root.left,key)
    return root
#Traversing the binary search tree
def inOrder(root):
    if root:
        inOrder(root.left)
        print(str(root.data)+" ",end=' ')
        inOrder(root.right)
        
root=Node(50)
root=insertBST(root,40)
root=insertBST(root,70)
root=insertBST(root,40)#checking for the duplicate element
root=insertBST(root,60)
root=insertBST(root,80)
print("the inorder traversal of bst is---")
inOrder(root)
print()
```

# Searching of an element in BST


```python
def searchBST(root,key):
    level=0
    if root is None:
        return False
    else:
        if root.data==key:
            return True
        elif root.data<key:
            return searchBST(root.right,key)
        else:
            return searchBST(root.left,key)
key=70
res=searchBST(root,key)
if res:
    print("Search element is found in the BST")
else:
    print(F"Search element {key} is not found in the BST")
```

# to know at which level the search key is found

```python
def searchBST(root,key,level):
    
    if root is None:
        return False
    else:
        if root.data==key:
            return level
        elif root.data<key:
            return searchBST(root.right,key,level+1)
        else:
            return searchBST(root.left,key,level+1)
key=80
level=1
res=searchBST(root,key,level)
if res:
    print(f"Search element {key} is found in the BST at level, {res}")
else:
    print(F"Search element {key} is not found in the BST")
```
