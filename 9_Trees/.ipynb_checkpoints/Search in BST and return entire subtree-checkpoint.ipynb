```python
class Node():
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
    def searchBST(self, root, val):
        if root is None:
            return
        if root.val==val:
            return root
        elif root.val<val:
            return self.searchBST(root.right,val)
        else:
            return self.searchBST(root.left,val)
        
root=Node(4)
root.left=Node(2)
root.right=Node(7)
root.left.left=Node(1)
root.left.right=Node(3)
obj=Node(0)
node=obj.searchBST(root,4)
arr=[]
if node:
   
    arr.append(node.val)
    arr.append(node.left.val) if node.left else None  # 60
    arr.append(node.right.val) if node.right else None  # 80
    print(arr)
else:
    print(arr)
```
