# T.C:O(n)

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
class LinkedList:
    def __init__(self):
        self.head=None
    def insertAtPos(self,pos,new_data):
        new_node=Node(new_data)
        temp=self.head
        count=0
        if pos==0:
            new_node.next=self.head
            self.head=new_node
            return
        while temp is not None and count<pos-1:
            temp=temp.next
            count +=1
        if temp is None:
            print("index out of range")
            return
        else:
            new_node.next=temp.next
            temp.next=new_node
    def printList(self):
        temp=self.head
        while temp:
            print(str(temp.data)+" ",end=" ")
            temp=temp.next
            
    def countNodes(self):
        temp=self.head
        count=0
        while temp:
            count +=1
            temp=temp.next
        return count
            
        
        
        
#driver code
llist=LinkedList()
llist.insertAtPos(0,10)
llist.insertAtPos(1,20)
llist.insertAtPos(2,30)
llist.insertAtPos(3,40)
llist.insertAtPos(4,50)
llist.insertAtPos(0,10)
llist.insertAtPos(1,20)
llist.insertAtPos(2,30)
llist.insertAtPos(3,40)
llist.insertAtPos(4,50)
llist.printList()
res_count=llist.countNodes()
print("\n count is--",res_count)
```
