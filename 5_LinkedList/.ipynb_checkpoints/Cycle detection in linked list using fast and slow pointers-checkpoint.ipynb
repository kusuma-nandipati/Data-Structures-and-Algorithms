```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
class LinkedList:
    def __init__(self):
        self.head=None
    def insertAtBeginning(self,new_data):
        new_node=Node(new_data)
        new_node.next=self.head
        self.head=new_node
    #function to detect cycle in the list
    def detectcycle(self):
        tortoise=self.head
        hare=self.head
        #moving fast pointers two step where as tortoise(slow pointer ) one step
        while tortoise and hare and hare.next:
            hare=hare.next.next
            tortoise=tortoise.next
            if tortoise==hare:
                return True
        return False
#driver code
llist=LinkedList()
llist.insertAtBeginning(50)
llist.insertAtBeginning(40)
llist.insertAtBeginning(30)
llist.insertAtBeginning(20)
llist.insertAtBeginning(10)
#creating cycle in linkedlist
llist.head.next.next.next.next=llist.head
#printing out the returned result
if llist.detectcycle():
    print("loop exists in the list")
else:
    print("no loop exists in the linkedlist")
```
