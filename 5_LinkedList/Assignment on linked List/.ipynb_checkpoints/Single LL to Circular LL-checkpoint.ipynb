```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
class LinkedList:
    def __init__(self):
        self.head=None
    def insertAtBeginning(self,new_Data):
        new_node=Node(new_Data)
        new_node.next=self.head
        self.head=new_node
    def convertToCLL(self):
        temp=self.head
        while temp:
            temp=temp.next
        temp=self.head
    def printList(self):
        temp=self.head
        while temp:
            print(str(temp.data)+" ",end=" ")
            temp=temp.next
llist=LinkedList()
llist.insertAtBeginning(50)
llist.insertAtBeginning(40)
llist.insertAtBeginning(30)
llist.insertAtBeginning(20)
llist.insertAtBeginning(10)
llist.printList()
llist.convertToCLL()
llist.printList()
```
