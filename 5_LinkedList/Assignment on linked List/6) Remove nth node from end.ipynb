# Approach-2 :Two pointers

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
        
class LinkedList:
    def __init__(self):
        self.head=None
    def insertionAtBeginning(self,new_data):
        new_node=Node(new_data)
        temp=self.head
        new_node.next=self.head
        self.head=new_node
        
        
    #method to delete nth node from list
    def removeNthnode(self,n):
        if n==0:
            return
        count=0
        dummy=Node(0)
        dummy.next=self.head
        temp=self.head
        
        '''counting no.of nodes in a linkedlist'''
        
        while temp:
            count +=1
            temp=temp.next
        steps_required=count-n
        temp=dummy
        
        '''number of steps required''' 
        
        while steps_required>0:
            steps_required -=1
            temp=temp.next
        temp.next=temp.next.next
        self.head=dummy.next
        
        '''printing the linkedlist'''
        
    def printList(self):
        temp=self.head
        while temp:
            print(str(temp.data)+" ",end=" ")
            temp=temp.next
            
#driver code
llist=LinkedList()
llist.insertionAtBeginning(0)
llist.insertionAtBeginning(0)
llist.insertionAtBeginning(2)
llist.insertionAtBeginning(2)
llist.insertionAtBeginning(1)
llist.removeNthnode(0)
llist.printList()
```
