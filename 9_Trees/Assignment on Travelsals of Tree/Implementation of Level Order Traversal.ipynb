```python
class Node:
    def __init__(self,data):
        self.data=data
        self.left=None
        self.right=None
    def level_order(self,root):
        #'''Base case condition'''
        if root is None:
            return
        #'''initializing queue'''
        queue=[root]
        while queue :
            node = queue.pop(0)
            print(node.data, end=" ")
             #'''check for root left value'''
            if node.left:
                queue.append(node.left)
            #'''check for root right value'''
            if node.right:
                queue.append(node.right)
        
        
        
#driver code
obj=Node(0)
root=Node(10)
root.left=Node(20)
root.right=Node(30)
root.left.left=Node(40)
root.left.right=Node(50)
root.right.left=Node(60)
root.right.right=Node(70)
print("the level order traversal of given Tree is----")
obj.level_order(root)
```
